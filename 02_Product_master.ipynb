```python
#Check product data
print("Total Products:", product_df.count())

product_df.printSchema()
```

```python
#Validate product records
# Check duplicate product IDs

product_df.groupBy("product_id") \
    .count() \
    .filter(F.col("count") > 1) \
    .show()

# Check invalid prices

product_df.filter(F.col("unit_price") <= 0).show()
```

```python
#Save Product Master in Delta Lake
product_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("product_master")
```
