```python
#Create the ActionCatalog table
from pyspark.sql import functions as F

action_data = [
    ("ACTN-2026-000001", "VIP_REWARD", "Champions", "LOYALTY", "APP", 500, "ACTIVE"),
    ("ACTN-2026-000002", "LOYALTY_OFFER", "Loyal", "MARKETING", "EMAIL", 200, "ACTIVE"),
    ("ACTN-2026-000003", "WELCOME_OFFER", "New Customers", "MARKETING", "EMAIL", 100, "ACTIVE"),
    ("ACTN-2026-000004", "RETENTION_OFFER", "At Risk", "MARKETING", "EMAIL", 250, "ACTIVE"),
    ("ACTN-2026-000005", "WINBACK_CAMPAIGN", "Hibernating", "MARKETING", "EMAIL", 150, "ACTIVE"),
    ("ACTN-2026-000006", "PRODUCT_RECOMMENDATION", "Developing", "MARKETING", "APP", 100, "ACTIVE"),
    ("ACTN-2026-000007", "SUPPORT_FOLLOWUP", "At Risk", "SERVICE", "PHONE", 50, "ACTIVE"),
    ("ACTN-2026-000008", "SUPPORT_FOLLOWUP", "Hibernating", "SERVICE", "PHONE", 50, "ACTIVE")
]

action_df = spark.createDataFrame(
    action_data,
    [
        "action_id",
        "action_name",
        "target_segment",
        "action_type",
        "channel",
        "estimated_cost",
        "action_status"
    ]
)

action_df = (
    action_df
    .withColumn("requires_marketing_consent",
        F.when(F.col("action_type") == "MARKETING", True).otherwise(False))
    .withColumn("created_date", F.lit("2026-10-30").cast("date"))
)

display(action_df)
```

```python
#Save the table
action_df.write.format("delta").mode("overwrite").saveAsTable("action_catalog")

print("Action Catalog created successfully!")
```

```python
#Check the actions
spark.table("action_catalog").groupBy(
    "target_segment"
).count().show()
```
