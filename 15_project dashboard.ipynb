```python
from pyspark.sql import functions as F

segment_df = spark.table("gold_segment_summary")
spending_df = spark.table("gold_customer_spending")
recommendation_df = spark.table("gold_recommendation_performance")
engagement_df = spark.table("gold_engagement_summary")

print("Gold tables loaded successfully!")
```

```python
#View 1 – Customer Segments
segment_df.createOrReplaceTempView("vw_customer_segments")
```

```python
#View 2 – Customer Spending
spending_df.createOrReplaceTempView("vw_customer_spending")
```

```python
#View 3 – Recommendation Performance
recommendation_df.createOrReplaceTempView("vw_recommendation_performance")
```

```python
#View 4 – Customer Engagement
engagement_df.createOrReplaceTempView("vw_customer_engagement")
```

```python
#Create a business KPI view
spark.sql("""
CREATE OR REPLACE TEMP VIEW vw_business_kpis AS
SELECT
    (SELECT SUM(total_customers)
     FROM vw_customer_segments) AS total_customers,

    (SELECT SUM(total_recommendations)
     FROM vw_recommendation_performance) AS total_recommendations,

    (SELECT SUM(converted_recommendations)
     FROM vw_recommendation_performance) AS converted_recommendations,

    (SELECT SUM(total_revenue)
     FROM vw_customer_spending) AS total_revenue
""")
```
